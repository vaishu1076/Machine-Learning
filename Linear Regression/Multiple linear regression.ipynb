##Multiple Linear Regression

Business -> Build a model that can predict accurate profits on basis of RND and MKT


Step1: Data Ingestion/Data Gathering

```python
import pandas as pd
path=r"https://raw.githubusercontent.com/sindhura-nk/Datasets/refs/heads/main/50_Startups.csv"
df = pd.read_csv(path)
df.head()
```

Step3: Separate the X and Y features

    X: RND , MKT
    Y: PROFIT

```python
X=df[['RND','MKT']]
Y=df[['PROFIT']]
display(X)
display(Y)
```

Step4: Feature Engineering:-> Data Cleaning

```python
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer

num_pipe = (SimpleImputer(strategy='median')).set_output(transform='pandas')
num_pipe
```

Step5: Separate our data into training and testing

dataset size is small: 80% and above training : 20% and below testing

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest= train_test_split(X,Y ,train_size=0.8,random_state=21)
```

```python
print(xtrain.shape,xtest.shape)
```

```python
print(ytrain.shape,ytest.shape)
```

Apply the pipeline on training and testing data

```python
num_pipe.fit(xtrain)
```


    Pipeline(steps=[('simpleimputer', SimpleImputer(strategy='median'))])

```python
xtrain_pre=num_pipe.transform(xtrain)
xtest_pre=num_pipe.transform(xtest)
```

Step5: Model Building

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(xtrain_pre,ytrain)
```

```python
ypred_test=model.predict(xtest_pre)
ypred_test[:5]
```

```python
ytest.head()
```

Step 6: Model Evaluation : Performance Metrics

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score

mse = mean_squared_error(ytest,ypred_test)
mae = mean_absolute_error(ytest,ypred_test)
rmse = mse**0.5
r2 = r2_score(ytest,ypred_test)

print(f"Testing scores")
print(f"MSE:{mse}")
print(f"MAE:{mae}")
print(f"RMSE:{rmse}")
print(f"R2 score-Coefficient of Determination:{r2*100:.2f}")
```

R2 Score for the model is above 80%. We can consider this model for final deployments.
